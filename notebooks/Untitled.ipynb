```python
import json
from pathlib import Path

# ========================================
# CONFIGURATION : Modifiez ces chemins
# ========================================
FICHIER_ENTREE = r"C:\Users\touad\Projects-Openclassrooms\Projet-Future-Vision-Transport\notebooks\p08_david_scanu_notebook_MobileNetV2_UNet (1).ipynb"
FICHIER_SORTIE = r"C:\Users\touad\Projects-Openclassrooms\Projet-Future-Vision-Transport\notebooks\Notebook_scanu.ipynb"

def extraire_code_notebook(fichier_entree, fichier_sortie):
    """
    Extrait uniquement les cellules de code d'un notebook Jupyter
    et les sauvegarde dans un nouveau notebook.
    
    Args:
        fichier_entree: Chemin vers le notebook source (.ipynb)
        fichier_sortie: Chemin vers le nouveau notebook (.ipynb)
    """
    try:
        # Lire le notebook source
        print(f"📖 Lecture du fichier: {fichier_entree}")
        with open(fichier_entree, 'r', encoding='utf-8') as f:
            notebook = json.load(f)
        
        # Extraire uniquement les cellules de code
        cellules_code = []
        for cellule in notebook.get('cells', []):
            if cellule.get('cell_type') == 'code':
                # Créer une nouvelle cellule de code sans les outputs
                nouvelle_cellule = {
                    'cell_type': 'code',
                    'execution_count': None,
                    'metadata': {},
                    'source': cellule.get('source', []),
                    'outputs': []
                }
                cellules_code.append(nouvelle_cellule)
        
        # Créer le nouveau notebook avec uniquement le code
        nouveau_notebook = {
            'cells': cellules_code,
            'metadata': notebook.get('metadata', {}),
            'nbformat': notebook.get('nbformat', 4),
            'nbformat_minor': notebook.get('nbformat_minor', 5)
        }
        
        # Sauvegarder le nouveau notebook
        with open(fichier_sortie, 'w', encoding='utf-8') as f:
            json.dump(nouveau_notebook, f, ensure_ascii=False, indent=1)
        
        print(f"✓ Extraction réussie!")
        print(f"  - Cellules de code extraites: {len(cellules_code)}")
        print(f"  - Fichier sauvegardé: {fichier_sortie}")
        
    except FileNotFoundError:
        print(f"✗ Erreur: Le fichier '{fichier_entree}' n'existe pas.")
        print(f"   Vérifiez le chemin dans la variable FICHIER_ENTREE")
    except json.JSONDecodeError:
        print(f"✗ Erreur: Le fichier '{fichier_entree}' n'est pas un JSON valide.")
    except Exception as e:
        print(f"✗ Erreur inattendue: {e}")

if __name__ == "__main__":
    # Si FICHIER_SORTIE n'est pas défini, générer automatiquement le nom
    if not FICHIER_SORTIE or FICHIER_SORTIE == "code_extrait.ipynb":
        chemin = Path(FICHIER_ENTREE)
        fichier_sortie_auto = chemin.stem + "_code_only" + chemin.suffix
    else:
        fichier_sortie_auto = FICHIER_SORTIE
    
    extraire_code_notebook(FICHIER_ENTREE, fichier_sortie_auto)
```
